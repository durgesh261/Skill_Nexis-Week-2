## 📦 Step 0: Imports & Environment Setup

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure display options
pd.set_option('display.max_columns', 20)
pd.set_option('display.width', 1000)
print('Libraries imported successfully!')
```

## 🔹 Question 1: Load a CSV file using Pandas and display basic info

```python
data_path = os.path.join('..', 'data', 'chocolate_sales.csv')
df = pd.read_csv(data_path)

print(f'DataFrame Dimensions: {df.shape[0]} rows, {df.shape[1]} columns\n')
df.head()
```

```python
# Display structure and datatypes
df.info()
```

```python
# Summary statistics for numerical variables
df.describe()
```

## 🔹 Question 2: Handle missing values and duplicates using Pandas

```python
# Check missing values
print('Missing Values Count per Column:')
print(df.isnull().sum())

# Check duplicate rows
print(f'\nTotal Duplicate Rows: {df.duplicated().sum()}')
```

```python
# Cleaning Strategy:
# 1. Impute missing Team names with 'Unassigned'
df_cleaned = df.copy()
df_cleaned['Team'] = df_cleaned['Team'].fillna('Unassigned')

# 2. Drop duplicates
initial_rows = len(df_cleaned)
df_cleaned = df_cleaned.drop_duplicates()

print(f'Missing values after imputation: {df_cleaned.isnull().sum().sum()}')
print(f'Rows retained: {len(df_cleaned)} of {initial_rows}')
```

## 🔹 Question 3: Group data by category and find total revenue

```python
category_revenue = df_cleaned.groupby('Category').agg(
    Total_Revenue=('Amount', 'sum'),
    Average_Order_Value=('Amount', 'mean'),
    Total_Boxes=('Boxes', 'sum'),
    Shipment_Count=('Amount', 'count')
).reset_index()

# Add revenue share percentage
total_rev = category_revenue['Total_Revenue'].sum()
category_revenue['Revenue_Share_%'] = (category_revenue['Total_Revenue'] / total_rev) * 100
category_revenue = category_revenue.sort_values(by='Total_Revenue', ascending=False)

category_revenue
```

## 🔹 Question 4: Sort data by multiple columns using Python

```python
# Sort 1: By Category (Ascending) and Amount (Descending)
sorted_cat_amount = df_cleaned.sort_values(
    by=['Category', 'Amount'],
    ascending=[True, False]
)

sorted_cat_amount[['Category', 'Product', 'Salesperson', 'Country', 'Amount', 'Boxes']].head(10)
```

```python
# Sort 2: By Country (Ascending) and SaleDate (Ascending)
sorted_geo_date = df_cleaned.sort_values(
    by=['Country', 'SaleDate'],
    ascending=[True, True]
)

sorted_geo_date[['Country', 'SaleDate', 'Product', 'Salesperson', 'Amount']].head(5)
```

## 🔹 Question 5: Create a correlation matrix for numerical columns

```python
numeric_columns = ['Amount', 'Customers', 'Boxes', 'Cost_per_box']
corr_matrix = df_cleaned[numeric_columns].corr()
corr_matrix
```

```python
# Visualizing Correlation Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.3f',
    linewidths=1,
    square=True,
    cbar_kws={'shrink': 0.8}
)
plt.title('Correlation Matrix — Awesome Chocolates Dataset', fontsize=14, pad=15, fontweight='bold')
plt.tight_layout()
plt.show()
```
